### 01.1 Series: değerler ve etiketler

#### Series nedir?
`Series`, her değerin bir indeks etiketiyle eşleştiği tek boyutlu yapıdır. NumPy dizisindeki
konumlara ek olarak anlamlı isimlerle çalışabiliriz. Üç ürünün stoklarını tek bir Series içinde tutabiliriz.

#### Değer ile indeks farklıdır
`pd.Series([8, 12], index=["Kalem", "Defter"])` ifadesinde sayılar stok, metinler ürün etiketidir.
`s.loc["Kalem"]` etiketle, `s.iloc[0]` konumla seçim yapar. İkisinin ayrıntısını 3. bölümde çalışacağız.
İndeks etiketlerinin her zaman benzersiz olacağı varsayılmaz; bu notebookun temel örneklerinde benzersizdir.

#### Ad ve veri tipi
`name`, Series’in neyi ölçtüğünü anlatan addır; indeks etiketi değildir. `dtype` saklanan veri tipidir.
`s.size` eleman sayısını verir. Özelliklerde parantez yoktur; `s.size()` yazmayız.


#### Çözümlü örnek — 01A

**Soru:** Üç ürünün stoklarını ürün isimleriyle eşleştir.

Önce hazır veriyi incele; ardından aşağıdaki kodu ve sonucunu birlikte oku.


````python
stok = pd.Series([8, 12, 5], index=["Kalem", "Defter", "Silgi"], name="stok")
display(stok)
print("Kalem stoku:", stok.loc["Kalem"])
print("Eleman sayısı:", stok.size)
````

Series üç kayıt içerir. Kalem etiketi 8 değerine karşılık gelir. Etiketler sayısal konumların yerine anlamlı seçim yapmamızı sağlar.

### 02.2 describe: ortalama ve dağılımı okumak

#### Hangi sütunu özetliyorsun?
`df["puan"].describe()` bir sayısal sütunun sayısını, ortalamasını, standart sapmasını, minimumunu,
çeyreklerini ve maksimumunu verir. Metin sütunlarının özeti farklıdır; burada sayısal sütunu açıkça seçiyoruz.

#### Özet satırlarını yorumlamak
`count` boş olmayan gözlem sayısıdır. `50%` medyandır. `25%` ve `75%` dağılımın çeyrekleridir.
`mean()` ortalamayı, `median()` medyanı hesaplar. Çok büyük bir gözlem ortalamayı medyana göre daha fazla etkileyebilir.

#### NumPy ile küçük bir fark
Pandas `Series.std()` varsayılan olarak `ddof=1` kullanır; NumPy `np.std()` varsayılanı `ddof=0` idi.
Aynı sonucu karşılaştırırken aynı `ddof` değerini seç. Bu örnekte bütün değerleri betimlemek için `std(ddof=0)` yazıyoruz.
Sayısal özetler ölçümün birimini ve hangi kayıtların hesaba girdiğini açıklamadan tek başına yeterli değildir.


#### Çözümlü örnek — 02B

**Soru:** Dört teslimat süresinin özetini incele.

Önce hazır veriyi incele; ardından aşağıdaki kodu ve sonucunu birlikte oku.


````python
df = pd.DataFrame({"sure_dk": [10, 20, 30, 100]})
display(df["sure_dk"].describe())
print("Ortalama:", df["sure_dk"].mean())
print("Medyan:", df["sure_dk"].median())
print("Standart sapma (ddof=0):", df["sure_dk"].std(ddof=0))
````

**Çözümün mantığı ve sonucu**

Ortalama 40, medyan 25 dakikadır. 100 dakikalık teslimat ortalamayı yükseltir. Özet, nedenini açıklamaz; ilgili kaydı ayrıca incelemek gerekir.

### 04.1 Boolean maske ve birleşik koşullar

#### Karşılaştırma bir maske üretir
`df["puan"] >= 70` her satır için True veya False içeren bir Series verir.
`df.loc[maske]` yalnızca True olan satırları seçer. NumPy’deki maskeleme mantığını etiketli tabloda kullanıyoruz.

#### İki koşulu birleştirmek
`&` iki koşulun da sağlanmasını, `|` en az birinin sağlanmasını ister. `~` maskeyi tersine çevirir.
Her karşılaştırmayı kendi parantezine al: `(kosul1) & (kosul2)`. Series üzerinde Python’un `and`/`or` sözcüklerini kullanma.

#### Önce maske, sonra sütunlar
`df.loc[maske, ["isim", "puan"]]` satır filtresini ve sütun seçimini birlikte yapar.
Filtreleme indeksi kendiliğinden sıfırlamaz; kalan etiketler kaynak kayıtları takip etmeye yardım eder.
Kaynağı değiştirmek istiyorsan ayrıca atama gerekir; burada yalnızca seçim yapıyoruz.


#### Çözümlü örnek — 04A

**Soru:** Puanı en az 70 olan ve Ankara’da yaşayan öğrencileri seç.

Önce hazır veriyi incele; ardından aşağıdaki kodu ve sonucunu birlikte oku.


````python
df = pd.DataFrame({"isim": ["Ece", "Ali", "Can"], "sehir": ["Ankara", "İzmir", "Ankara"], "puan": [80, 90, 60]})
maske = (df["puan"] >= 70) & (df["sehir"] == "Ankara")
display(maske)
display(df.loc[maske, ["isim", "puan"]])
````

Yalnızca Ece iki koşulu birlikte sağlar. Ali yüksek puanlı olsa da Ankara koşulunu sağlamaz.

### 05.2 loc ile hedefi güncellemek, copy ile ayırmak

yayılmasına güvenilmez; zincirleme atama hedef tabloyu güncellemez. Bu notebook hem güncel davranışa

### 06.1 set_index ve reset_index

#### Kimliği indeks yapmak
Bir ürün kodu kaydı tanımlıyorsa `df.set_index("kod")` ile onu indeks yapabiliriz.
Bu sütun varsayılan olarak normal sütunlar arasından çıkar; değerleri satır etiketi olur.
Sonuçtaki `index.is_unique` özelliği etiketlerin tekil olup olmadığını söyler.
Tekil kimlik bekliyorsan bunu kontrol ederek tekrar durumunda açık hata verebilirsin.

#### Sonucu saklamak
Bu notebookta işlemlerin sonucunu yeni değişkene atıyoruz: `kodlu = df.set_index("kod")`.
Yalnızca metodu çağırmak kaynak tablonun indeksini değiştirmez. `inplace=True` kullanmak zorunda değilsin.

#### Tekrar sütuna dönmek
`reset_index()` mevcut indeksi sütuna taşır ve 0’dan başlayan indeks oluşturur.
`reset_index(drop=True)` ise eski indeksi sütuna eklemez. Eski indeks bir kayıt kimliği ise
onu kaybetmenin uygun olup olmadığını önce düşün. Filtre sonrası düzenli numaralar istenirse drop=True işe yarar.


#### Çözümlü örnek — 06A

````python
df = pd.DataFrame({"kod": ["K01", "D01"], "urun": ["Kalem", "Defter"], "fiyat": [15, 40]})
kodlu = df.set_index("kod")
if not kodlu.index.is_unique:
    raise ValueError("Ürün kodları tekil olmalı.")
display(kodlu)
print(kodlu.loc["D01", "fiyat"])
display(kodlu.reset_index())
````

### 10.2 agg ile birden çok özet üretmek

#### Tek raporda birden çok ölçü
Bir kategori için yalnızca toplam değil ortalama, maksimum ve gözlem sayısı da gerekebilir.
`agg`, her gruba bu işlemleri uygulayıp bir özet tablosu üretir.

#### Adlandırılmış toplulaştırma
`toplam=("tutar", "sum")` ifadesinde soldaki ad çıktı sütunu, sağdaki tuple ise
hangi kaynak sütuna hangi işlemin uygulanacağını belirtir. Kaynak ile sonuç adını birbirinden ayırır.
`as_index=False` grup anahtarını normal bir sütun olarak tutar; raporu okumayı kolaylaştırır.

#### Eksik değer politikasını belirtmek
Standart sum bütün değerler eksikse sıfır verebilir. Bilinmeyeni korumak için bu örnekte
toplam fonksiyonunu `lambda s: s.sum(min_count=1)` olarak açıkça seçiyoruz.
`count` dolu gözlemleri, `size` satırları sayar; raporundaki sayının hangisi olduğunu adından belli et.


#### Çözümlü örnek — 10B

**Soru:** Her kategorinin toplamını, ortalamasını ve kayıt sayısını tek tabloda göster.

Önce hazır veriyi incele; ardından aşağıdaki kodu ve sonucunu birlikte oku.


````python
df = pd.DataFrame({"kategori": ["A", "A", "B"], "tutar": [20, 40, 90]})
rapor = df.groupby("kategori", as_index=False).agg(
    toplam=("tutar", lambda s: s.sum(min_count=1)),
    ortalama=("tutar", "mean"),
    kayit_sayisi=("tutar", "size"),
)
display(rapor)
````

#### Boş hücre ile sıfırı ayır
Bir öğrenci için ders kaydı yoksa oluşan hücre eksiktir; bu otomatik olarak sıfır puan değildir.
`fill_value=0` ancak sıfır anlamlıysa kullanılmalıdır. `pivot_table` varsayılan özeti mean olsa da

### 13.2 melt ile sütunları satırlara taşımak

#### Genişten uzuna dönmek
`melt`, ölçüm sütunlarını bir değişken adı sütunu ve bir değer sütunu içinde toplar.
Örneğin ocak ve subat sütunlarını ay ile tutar sütunlarına dönüştürebiliriz.

#### Kimlikleri korumak
`id_vars` sabit kalacak kimlik sütunlarını, `value_vars` satırlara taşınacak ölçümleri belirtir.

### 14.2 resample ve rolling ile zaman özeti

#### Dönemlere ayırmak
Tarih indeksli bir Series’de `resample("MS")` ay başlangıçlarıyla etiketlenmiş aylık gruplar kurar.
Ardından sum veya mean gibi işlemi seçersin. Burada `.sum(min_count=1)` ile hiç ölçümü olmayan dönemi
otomatik sıfır saymıyoruz. Önce geçersiz tarihleri ele al ve tarih sırasını kontrol et.

#### Hareketli pencere
`rolling(window=2, min_periods=2).mean()` her konumda son iki gözlemin ortalamasını hesaplar.
İlk gözlemde iki kayıt olmadığı için sonuç eksiktir. Pencere bir ortalama eğilimini görmeyi sağlar;
verinin kısa süreli dalgalanmalarını yumuşatır.

#### İki gözlem ile iki gün aynı şey değildir
`window=2` kayıt sayısına bakar. Zaman indeksinde `rolling("2D")` ise zaman aralığına bakar.
Gün atlanan veya bir günde birden çok kayıt bulunan veride sonuçlar farklılaşır.
Zaman analizi öncesi `sort_index()` ile sıralamayı açıkça yapıyoruz.


#### Çözümlü örnek — 14B

**Soru:** Harcama serisinin aylık toplamını ve iki gözlemlik hareketli ortalamasını bul.

Önce hazır veriyi incele; ardından aşağıdaki kodu ve sonucunu birlikte oku.


````python
tarih = pd.to_datetime(["2026-01-01", "2026-01-02", "2026-02-01", "2026-02-02"])
s = pd.Series([10, 20, 30, 50], index=tarih, name="tutar").sort_index()
display(s.resample("MS").sum(min_count=1))
display(s.rolling(window=2, min_periods=2).mean())
````

**Çözümün mantığı ve sonucu**

Ocak toplamı 30, şubat toplamı 80 olur. İki gözlemli ortalamalar ilk satırda eksik, sonra 15, 25 ve 40 olur.

### 17.2 Kategori harcamasını bütçeyle karşılaştırmak

#### Önce aynı düzeye getir
İşlem düzeyindeki harcama ile kategori düzeyindeki bütçeyi doğrudan satır satır karşılaştırma.
Önce groupby ile kategori toplamlarını üret; sonra kategori anahtarı üzerinden bütçeyi ekle.

#### Eşleşmeyi doğrula
Özet tabloda ve bütçe tablosunda her kategoriden bir satır bekliyoruz; `validate="one_to_one"` bunu denetler.
`how="left"` harcama kategorilerini korur. Bütçede bulunmayan kategori varsa bütçe eksik kalır;
bu durum "bütçe sıfır" diye yorumlanmaz.

#### Kullanım oranı
`harcama / butce` 1 ise bütçenin tamamı kullanılmıştır; 1’den büyükse aşım vardır.
Sıfır veya negatif bütçede bu oranı hesaplamıyoruz. Bu left join raporu yalnızca harcama yapılan
kategorileri kapsar; bütçesi olup harcaması olmayan kategorileri de göstermek istersen kapsamı outer join ile genişletebilirsin.


#### Çözümlü örnek — 17B

**Soru:** Harcama yapılan kategorilerin bütçe kullanım oranlarını bul.

Önce hazır veriyi incele; ardından aşağıdaki kodu ve sonucunu birlikte oku.


````python
gider = pd.DataFrame({"kategori": ["Yemek", "Ulaşım", "Yemek", "Market"], "tutar": [100, 40, 60, 200]})
butce = pd.DataFrame({"kategori": ["Yemek", "Ulaşım", "Market"], "butce": [150, 100, 300]})
ozet = gider.groupby("kategori", as_index=False).agg(harcama=("tutar", "sum"))
rapor = ozet.merge(butce, on="kategori", how="left", validate="one_to_one")
rapor["kullanim_orani"] = rapor["harcama"] / rapor["butce"].where(rapor["butce"] > 0)
display(rapor.sort_values("kullanim_orani", ascending=False))
````

**Çözümün mantığı ve sonucu**

Yemek 160/150 ile bütçesini aşmıştır. Ulaşım 0.4, Market yaklaşık 0.667 oranındadır. Oranlar yüzde değildir; yüzde için 100 ile çarpılır.


#### Sen çöz — 17B

**Amaç:** Birleştirmesi hazır rapora kullanim_orani sütununu ekle. Yalnızca pozitif bütçelerde bölme yap.

**Yapacağın iş:** Hazır verinin altındaki `sonuc = None` satırını kendi çözümünle değiştir.
Gerekirse önce bir ara değişken oluştur, en son sonucu `sonuc` içine kaydet.

**Hedef çıktı:** Kullanım oranları [0.8, 0.4, eksik].

<details><summary>İpucu — gerekirse aç</summary>

</details>

## 18. ML köprüsü: tutarlı veri hazırlama

### 18.1 Doldurma ve standardizasyonu eğitimden öğrenmek

#### Eğitim ile testin görevi farklıdır
Model eğitim verisinden öğrenir. Test verisi daha önce görülmeyen veriye dair değerlendirme içindir.
Doldurma medyanı ve ölçek gibi hazırlık parametrelerini tüm veriden hesaplamak testten bilgi sızdırabilir.
Bu bölüm model kurmaz; yalnızca hazırlık ayrımını gösterir. Önceki dersler bitmeden çalışmak zorunda değilsin.

#### fit ve transform düşüncesi
Eğitim medyanını hesapla, eğitim eksiklerini onunla doldur. Ortalama ve standart sapmayı doldurulmuş
eğitimden hesapla. Test eksiklerini aynı medyanla doldurup aynı ortalama ve sapmayla dönüştür.
Burada eğitim/test tabloları önceden ayrılmıştır; gerçek veride ayırma stratejisi zaman ve grup yapısına bağlıdır.

#### Sabit sütun ve tamamen eksik sütun
Eğitim sapması sıfırsa bu örnekte bölende 1 kullanırız; böylece eğitimdeki sabit değerler sıfıra dönüşür.
Eğitim sütunu tamamen eksikse medyan öğrenilemez; bu durum için ayrı politika gerekir.
Örnekte bunu kontrol ediyoruz. Sayısal sonsuzlukları da bu hazırlıkta eksik olarak ele aldığımızı açıkça belirtiyoruz.


#### Çözümlü örnek — 18A

**Soru:** Test verisini yalnızca eğitimden hesaplanan değerlerle hazırla.

Önce hazır veriyi incele; ardından aşağıdaki kodu ve sonucunu birlikte oku.


````python
egitim = pd.Series([10.0, np.nan, 30.0], name="olcum")
test = pd.Series([np.nan, 50.0], name="olcum")
egitim = egitim.where(np.isfinite(egitim))
test = test.where(np.isfinite(test))
medyan = egitim.median()
if pd.isna(medyan):
    raise ValueError("Eğitimde en az bir geçerli ölçüm olmalı.")
egitim_dolu = egitim.fillna(medyan)
ortalama = egitim_dolu.mean()
sapma = egitim_dolu.std(ddof=0)
olcek = sapma if sapma > 0 else 1.0
egitim_hazir = (egitim_dolu - ortalama) / olcek
test_hazir = (test.fillna(medyan) - ortalama) / olcek
display(egitim_hazir)
display(test_hazir)
````

**Çözümün mantığı ve sonucu**

Eğitim medyanı ve ortalaması 20 olur. Testin eksik ölçümü 20 ile doldurulur ve sıfıra dönüşür. Testteki 50, eğitim ortalamasını veya ölçeğini değiştirmez.


#### Sen çöz — 18A

**Amaç:** Testteki eksik değeri hazır eğitim medyanıyla doldur; hazır eğitim ortalaması ve ölçeğiyle standardize et.

**Yapacağın iş:** Hazır verinin altındaki `sonuc = None` satırını kendi çözümünle değiştir.
Gerekirse önce bir ara değişken oluştur, en son sonucu `sonuc` içine kaydet.

**Hedef çıktı:** [0.0, 2.0]

Önce `test.fillna(egitim_medyani)`, sonra `(dolu - egitim_ortalamasi) / egitim_olcegi` kullan.

</details>

### 18.2 get_dummies ve eğitim/test sütunlarını eşlemek

#### Kategoriyi ayrı gösterge sütunlarıyla temsil etmek
`pd.get_dummies(s, prefix="sehir", dtype=int)` her görülen kategori için 0/1 sütunu üretir.
Bu, Ankara=1, İzmir=2 gibi yapay bir büyüklük sırası vermeden kategoriyi temsil eder.
Gerçekten eksik kategoriyi saklamak için `dummy_na=True` düşünülebilir; bu örnekte bütün etiketler doludur.

#### Testte farklı kategoriler bulunabilir
Eğitim ve test için ayrı dummy üretince sütunlar farklı olabilir. Modelin beklediği eğitim şemasını
`test_kod.reindex(columns=egitim_kod.columns, fill_value=0)` ile uygula. Kolon sırasının da aynı olması gerekir.

#### Bilinmeyen kategori politikasını açık tut
Eğitimde görülmeyen kategori bu basit yöntemde eğitim sütunlarında tamamen sıfır olur.
Bu bir politikadır, kategori öğrenilmiş değildir. Önce `~test.isin(egitim.unique())` ile bilinmeyenleri raporla.
Bu örnekte hiçbir dummy sütununu düşürmüyoruz; gerçek ML akışında encodera ait fit/transform ve
bilinmeyen kategori ayarları bir pipeline içinde yönetilebilir. Bunlar sonraki ML aşamasının konusudur.


#### Çözümlü örnek — 18B

**Soru:** Testin kategori sütunlarını eğitimdeki şemayla eşleştir.

Önce hazır veriyi incele; ardından aşağıdaki kodu ve sonucunu birlikte oku.


````python
egitim = pd.Series(["Ankara", "İzmir", "Ankara"], name="sehir")
test = pd.Series(["İzmir", "Bursa"], name="sehir")
bilinmeyen = ~test.isin(egitim.unique())
display(test.loc[bilinmeyen])
egitim_kod = pd.get_dummies(egitim, prefix="sehir", dtype=int)
test_kod = pd.get_dummies(test, prefix="sehir", dtype=int)
test_hazir = test_kod.reindex(columns=egitim_kod.columns, fill_value=0)
display(egitim_kod)
display(test_hazir)
````
